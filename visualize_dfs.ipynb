```python
df = pd.read_json('./data/greif/temp/chamados_suporte_enduser_retorno_new.json')

df = df.join(pd.json_normalize(df['Status / Prazos']))
df.drop(columns='Status / Prazos', inplace=True)

df = df.join(pd.json_normalize(df['Protocolo - Funcionário - Matrícula ']))
df.drop(columns='Protocolo - Funcionário - Matrícula ', inplace=True)

df = df.join(pd.json_normalize(df['Assunto - Tópicos - Descrição']))
df.drop(columns='Assunto - Tópicos - Descrição', inplace=True)

df = df.join(pd.json_normalize(df['Redirecionamentos']))
df.drop(columns='Redirecionamentos', inplace=True)

df['Prazo'] = df['Prazo'].str.strip().str.split(' ').str[0].astype('Int64')

df['Data Abertura'] = pd.to_datetime(df['Data Abertura']).dt.floor('D')
df['Data Encerramento'] = df['Data Abertura'] + pd.to_timedelta(df['Prazo'], unit='D')

df.drop(columns=df.columns[0], inplace=True)
df

df.to_excel('ATESTADOS_GREIF_NOVO.xlsx')
df
```

```python
df = pd.read_excel('./data/rech/ATESTADOS_RECH_NOVO_NAO_TRATADO.xlsx')
# df.info()

# hasDigit = lambda string: df['Quantidade de Dias de Afastamento'].str.strip().str.split(' ').str[0].str.isdigit

df['Quantidade de Dias de Afastamento'] = pd.to_numeric(
    df['Quantidade de Dias de Afastamento'].str.strip().str.split(' ').str[0], 
    errors='coerce'
)
# df['Quantidade de Dias de Afastamento'] = df['Quantidade de Dias de Afastamento'].str.strip().str.split(' ').str[0].astype('Int64')\
#                                             if hasDigit(df['Quantidade de Dias de Afastamento'].str.strip().str.split(' ').str[0])\
#                                             else df['Quantidade de Dias de Afastamento']

df['Data Inicial do afastamento'] = pd.to_datetime(df['Data Inicial do afastamento'], errors='coerce', dayfirst=True).dt.floor('D')

# Calcular data fim afastamento
df['data fim afastamento'] = df.apply(
    lambda row: row['Data Inicial do afastamento'] + pd.to_timedelta(row['Quantidade de Dias de Afastamento'], unit='D')
    if pd.notna(row['Data Inicial do afastamento']) and pd.notna(row['Quantidade de Dias de Afastamento'])
    else pd.NaT,
    axis=1
)

df = df[df['Data Inicial do afastamento'] > pd.to_datetime('18/08/2025', dayfirst=True)]


# Normalizar hora e manter apenas a data
df['data fim afastamento'] = df['data fim afastamento'].dt.normalize()

# Se quiser mostrar no formato Y-m-d como string
df['data fim afastamento'] = df['data fim afastamento'].dt.strftime('%Y-%m-%d')
df['Data Inicial do afastamento'] = df['Data Inicial do afastamento'].dt.strftime('%Y-%m-%d')

# df = df[df['Data Inicial do afastamento'] > pd.to_datetime('19/09/2025', dayfirst=True)]
# df.drop(columns='data fim afastamento', inplace=True)



df.to_excel('teste.xlsx', index=False)
df
```
